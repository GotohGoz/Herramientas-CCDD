# Prueba con datos sintéticos

## Señal 1D

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

n = np.arange(100)
senal_demo = np.sin(2 * np.pi * n / 20)
tabla_demo = pd.DataFrame({"indice": n, "valor_sintetico": senal_demo})
print(tabla_demo.head())
plt.plot(tabla_demo["indice"], tabla_demo["valor_sintetico"])
plt.xlabel("Indice sintetico")
plt.ylabel("Amplitud sintetica (sin unidad fisica)")
plt.title("Ensayo de software: secuencia sintetica")
plt.tight_layout()
plt.show()
```

## Señal 2D

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
gris_demo = np.zeros((64, 64), dtype=np.uint8)
gris_demo[16:48, 16:48] = 180
imagen_demo = Image.fromarray(gris_demo)
print("Modo:", imagen_demo.mode, "tamano:", imagen_demo.size)
plt.imshow(imagen_demo, cmap="gray", vmin=0, vmax=255)
plt.title("Ensayo de software: imagen sintetica")
plt.axis("off")
plt.show()
```
